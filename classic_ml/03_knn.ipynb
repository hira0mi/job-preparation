```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer(as_frame=True)#попробуем использовать pd.DataFrame(), потому что с as_frame=True у меня не работает
df = pd.DataFrame(data.data, columns=data.feature_names)
df['Target'] = data.target
df.head()
```

Сначала попробую без скейлера

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
X = df.drop(columns=['Target'])
y = df['Target']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.3)#надо бы узнать, за что отвечает random_state и есть ли смысл ставить разные значения
model = KNeighborsClassifier()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
#я допустил data leakage? accuracy=0.9590643274853801
```

теперь со скейлером


```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)#не помню писал ли я это во втором ноутбуке, но если применить к тесту fit_transform(), то можно получить data leakage
model.fit(X_train, y_train)
y_pred_with_scaler = model.predict(X_test)
accuracy_with_scaler = accuracy_score(y_test, y_pred_with_scaler)
print(accuracy_with_scaler)#
```
