Градентный бустинг - ансамбль из простых моделей, которые, в отличие от RandomForest, исправляет ошибки предыдущих, из чего получается сильная модель

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../datasets/WineQT.csv')
df.info()
```

```python
X = df.drop(columns=['quality', 'Id'])
y = df['quality']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.3, stratify=y)
for n in [50, 100, 250, 500]:
    for depth in [10, 20, None]:
        model = RandomForestClassifier(n_estimators=n, max_depth=depth, random_state=42)
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        f1 = f1_score(y_test, y_pred, average='macro')
        print(f'n_estimators={n}, max_depth={depth}, f1_score:{f1}')
```

```python

for n in [50, 100, 250, 500]:
    for depth in [3, 5, 10]:
        for rate in [0.02, 0.05, 0.1, 0.15]:
            model = GradientBoostingClassifier(n_estimators=n, max_depth=depth, learning_rate=rate)
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            f1 = f1_score(y_test, y_pred, average='macro')
            print(f'n_estimators={n}, max_depth={depth}, learning_rate={rate} f1_score:{f1}')
```

На данном датасете Random Forest показал немного более высокий macro-F1, чем Gradient Boosting. Подбор гиперпараметров Gradient Boosting не позволил превзойти Random Forest. Результат показывает, что преимущество одного ансамблевого метода над другим не является универсальным и зависит от структуры данных и параметров моделей.
